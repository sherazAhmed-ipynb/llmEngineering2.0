### ***<span style= "color:#85BB65;"> Creating a Tensor </span>***

```python
# Empty function allocate a memory.
# Empty does not have any value it just asign a random location. The values that are shown are already there.
ten = torch.empty(2,3) # 2rows, 3columns
ten
```

```python
# Cheak type
type(ten)
```

```python
# Using zeros
torch.zeros(2,3)
```

```python
# Using rand
torch.rand(2,3)
```

```python
# Use of seed
torch.manual_seed(50)
torch.rand(2,3)
```

```python
# Using tensor
torch.tensor([[1,2,3],[4,5,6]])
```
