## ***<span style= "color:purple;">Imports </span>***

```python
import torch
```

### ***<span style= "color:#85BB65;"> Creating a simple autograde in pyTorch </span>***

```python
x = torch.tensor(3.0, requires_grad=True)
```

##### Here pytorch generates a computation graph internally.

x----------> sq -----------> y 

- x is the input as a tensor.
- sq is function.
- Which gave us y as the output.

```python
y = x**2
```

```python
x
```

```python
y
```

##### Now we calculate the drivative 

```python
y.backward()
```

```python
x.grad
```

### ***<span style= "color:#85BB65;"> Creating a nested function in pyTorch </span>***

```python
new_x = torch.tensor(3.0, requires_grad=True)
new_x
```

new_x----------> sq -----------> new_y -----------> sin ------------> new_z 

```python
new_y = new_x ** 2
new_y
```

```python
new_z = torch.sin(new_y)
new_z
```

```python
new_z.backward()
```

```python
new_x.grad
```

##### Here x is a leaf tnesor and pytorch only stores gradients by default for leaf tensors. Which means you can only generate the grad of x. y and z are non-leaf tensors.

```python
new_z.grad
```
