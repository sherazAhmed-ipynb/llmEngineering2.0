## ***<span style= "color:purple;">Imports </span>***

```python
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
df
```

### ***<span style= "color:#85BB65;">Train test split </span>***

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:,1:], df.iloc[:,0], test_size=0.2)
```

### ***<span style= "color:#85BB65;">Scaling </span>***

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

### ***<span style= "color:#85BB65;">Label encoding </span>***

```python
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
```

### ***<span style= "color:#15AB75;">Numpy arrays to Pytorch tensors </span>***

```python
X_train_tensor = torch.from_numpy(X_train)
X_test_tensor = torch.from_numpy(X_test)
y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)
```

## ***<span style= "color:darkred;">Defining model </span>***

```python
class mySimpleNN():
    def __init__(self, X):
        
        self.weights = torch.rand(X.shape[1],1, dtype=torch.float64, requires_grad=True)
        self.bais = torch.zeros(1, dtype=torch.float64, requires_grad=True )

    def forward(self, X):
        z = torch.matmul(X, self.weights) + self.bais
        y_pred = torch.sigmoid(z)
        return y_pred
        
```

```python
#Important Parameters
learning_rate = 0.1
epochs = 25
```

```python
#Training Pileline

#model creation
model = mySimpleNN(X_train_tensor)

#define loop
for epoch in range (epochs):

#fowrad pass
    y_pred = model.forward(X_train_tensor)
    print(y_pred)
```

## ***<span style= "color:darkred;">Defining model with the help of NN module </span>***

#### what is nn module?
The torch.nn in PyTorch is a core library that provides a wide array of classes and fuctions desgined to help developers build neural networks.

```python
# create model class
import torch
import torch.nn as nn

class Model(nn.Module):

    def __init__(self, num_featues):

        super().__init__()
        self.linear = nn.Linear(num_featues, 1)
        self.sigmoid = nn.Sigmoid()

    
    def forward(self, features):

        out = self.linear(features)
        out = self.sigmoid(out)
        return out
```

```python
# Neural network with hidden layer
import torch
import torch.nn as nn

class Model(nn.Module):

    def __init__(self, num_features):

        super().__init__()
        self.linear1 = nn.Linear(num_features,3)
        self.relu = nn.ReLU()
        self.linear2 = nn.Linear(3,1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, features):

        out = self.linear1(features)
        out = self.relu(out)
        out = self.linear2(out)
        out = self.sigmoid(out)
        
        return out

```

```python
# dataset
features = torch.rand(10,5)

model = Model(features.shape[1])

model(features)
```
